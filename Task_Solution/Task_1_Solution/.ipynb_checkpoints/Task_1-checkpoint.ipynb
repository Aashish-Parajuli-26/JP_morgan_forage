```python
df['Year'] = df['Dates'].dt.year
df['Month'] = df['Dates'].dt.month

pivot = df.pivot(index='Month', columns='Year', values='Prices')
pivot['Avg_YOY_Changes']=pivot.diff(axis=1).mean(axis=1)
pivot
```

why pivot table .. to see the year wise differences and calculate the avg YOY changes . as we can see starting jan its +0.56 
but beforethe end of the year this deferences solely rises to 0.6 . which indicates every year price rises as we move forward

From table we can see that month 8 that is august has the lowest avg increase and nov has highest increase of 6.3

```python
overall =pivot['Avg_YOY_Changes'].mean()
```

```python
last_known = pivot[[2020, 2021, 2022, 2023, 2024]].ffill(axis=1).iloc[:, -1]
```

```python
last_known
```

```python
pivot['blended_rate'] = (pivot['Avg_YOY_Changes']+overall)/2
```

```python
pivot
```

```python
pivot[2025] = last_known+pivot['blended_rate']
```

```python
pivot
```

```python
last_known = pivot[[2020, 2021, 2022, 2023, 2024]].ffill(axis=1).iloc[:, -1]
year_columns = [2020, 2021, 2022, 2023, 2024]
last_known_year = pivot[year_columns].apply(lambda row: row.last_valid_index(), axis=1)
def estimate(date_input):
    date_input=pd.to_datetime(date_input)
    month=date_input.month
    year=date_input.year

    last_price=last_known[month]
    anchor_year = last_known_year[month]
    blended_rate = pivot.loc[month, 'blended_rate']

    years_forward = year - anchor_year

    estimated = last_price + (blended_rate * years_forward)
    return estimated

print(estimate('2025-03-15'))
print(estimate('2026-01-01'))
print(estimate('2022-06-30'))  # test against a value you already know
print(estimate('2024-10-30'))
print(estimate('2024-09-30'))
```
